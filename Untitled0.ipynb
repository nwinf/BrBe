```python
# ================================================================
# HR POLICY AI PROJECT
# DOCUMENT PROCESSING AND TABLE RECONSTRUCTION
# ================================================================
# This cell performs:
# 1. Install/import required libraries
# 2. Upload HR Policy PDF
# 3. Extract text from all pages
# 4. Detect and extract tables
# 5. Clean extracted text
# 6. Inspect document statistics
# 7. Reconstruct the dress-code table on page 16
# ================================================================


# ------------------------------------------------
# STEP 1: INSTALL AND IMPORT LIBRARIES
# ------------------------------------------------

!pip -q install pymupdf pandas

import fitz
import pandas as pd
import re
import os
from collections import defaultdict
from google.colab import files

print("Libraries loaded successfully.")


# ------------------------------------------------
# STEP 2: UPLOAD THE HR POLICY PDF
# ------------------------------------------------

print("\nUpload your HR Policy PDF.")

uploaded = files.upload()

if not uploaded:
    raise ValueError("No PDF file was uploaded.")

pdf_filename = next(iter(uploaded.keys()))

if not pdf_filename.lower().endswith(".pdf"):
    raise ValueError("Please upload a PDF file.")
```

```python
pdf_document = fitz.open(pdf_filename)

print("\nDOCUMENT INFORMATION")
print("=" * 60)
print("Filename:", pdf_filename)
print("Total pages:", len(pdf_document))
print("File size:", round(os.path.getsize(pdf_filename) / 1024, 2), "KB")


# ------------------------------------------------
# STEP 3: EXTRACT TEXT AND TABLES FROM ALL PAGES
# ------------------------------------------------

documents = []
extracted_tables = []

for page_number, page in enumerate(pdf_document, start=1):

    # Extract page text
    page_text = page.get_text("text", sort=True)

    documents.append({
        "document_name": pdf_filename,
        "page_number": page_number,
        "text": page_text.strip(),
        "character_count": len(page_text.strip())
    })

    # Detect tables on the page
    table_finder = page.find_tables()

    for table_number, table in enumerate(
        table_finder.tables, start=1
    ):

        rows = table.extract()

        extracted_tables.append({
            "document_name": pdf_filename,
            "page_number": page_number,
            "table_number": table_number,
            "rows": rows,
            "row_count": len(rows),
            "column_count": max(
                (len(row) for row in rows),
                default=0
            )
        })


print("\nDOCUMENT EXTRACTION RESULTS")
print("=" * 60)
print("Pages processed:", len(documents))
print("Tables detected:", len(extracted_tables))

empty_pages = [
    doc["page_number"]
    for doc in documents
    if not doc["text"]
]

print("Empty pages:", len(empty_pages))
print("Empty page numbers:", empty_pages)


# ------------------------------------------------
# STEP 4: CLEAN EXTRACTED PAGE TEXT
# ------------------------------------------------

def clean_page_text(text):

    if not text:
        return ""

    # Normalize line endings
    text = text.replace("\r\n", "\n")
    text = text.replace("\r", "\n")

    # Remove unnecessary spaces and tabs
    text = re.sub(r"[ \t]+", " ", text)

    # Remove spaces around line breaks
    text = re.sub(r" *\n *", "\n", text)

    # Reduce excessive blank lines
    text = re.sub(r"\n{3,}", "\n\n", text)

    return text.strip()


cleaned_documents = []

for doc in documents:

    cleaned_text = clean_page_text(doc["text"])

    cleaned_documents.append({
        "document_name": doc["document_name"],
        "page_number": doc["page_number"],
        "original_text": doc["text"],
        "cleaned_text": cleaned_text,
        "original_character_count": len(doc["text"]),
        "cleaned_character_count": len(cleaned_text)
    })


original_total = sum(
    doc["original_character_count"]
    for doc in cleaned_documents
)

cleaned_total = sum(
    doc["cleaned_character_count"]
    for doc in cleaned_documents
)

reduction = (
    ((original_total - cleaned_total) / original_total) * 100
    if original_total else 0
)

print("\nTEXT CLEANING RESULTS")
print("=" * 60)
print("Original characters:", original_total)
print("Cleaned characters:", cleaned_total)
print("Character reduction:", round(reduction, 2), "%")


# ------------------------------------------------
# STEP 5: INSPECT A SAMPLE OF EXTRACTED TEXT
# ------------------------------------------------

print("\nSAMPLE: PAGE 6")
print("=" * 60)

print(cleaned_documents[5]["cleaned_text"][:3000])

print("\nSAMPLE: PAGE 7")
print("=" * 60)

print(cleaned_documents[6]["cleaned_text"][:3000])


# ------------------------------------------------
# STEP 6: INSPECT DETECTED TABLES
# ------------------------------------------------

print("\nTABLE DETECTION SUMMARY")
print("=" * 60)

table_summary = pd.DataFrame([
    {
        "page_number": table["page_number"],
        "table_number": table["table_number"],
        "rows": table["row_count"],
        "columns": table["column_count"]
    }
    for table in extracted_tables
])

display(table_summary)

print("\nTABLES DETECTED ON PAGE 16")
print("=" * 60)

for table in extracted_tables:

    if table["page_number"] == 16:

        print("\nTable number:", table["table_number"])

        for row in table["rows"]:
            print(row)


# ------------------------------------------------
# STEP 7: SELECT PAGE 16 FOR RECONSTRUCTION
# ------------------------------------------------

if len(pdf_document) < 16:
    raise ValueError("The PDF does not contain page 16.")

page_16 = pdf_document[15]

print("\nPAGE 16 TEXT")
print("=" * 60)

print(page_16.get_text("text", sort=True))


# ------------------------------------------------
# STEP 8: IDENTIFY MEN AND WOMEN TABLE SECTIONS
# ------------------------------------------------

def identify_attire_sections(page):

    words = page.get_text("words", sort=True)

    men_positions = []
    women_positions = []

    for word in words:

        x0, y0, x1, y1, text = word[:5]

        normalized = text.strip().lower()

        # Restrict search to the known attire table area
        if not (80 <= y0 <= 400):
            continue

        if normalized == "men":
            men_positions.append(y0)

        elif normalized == "women":
            women_positions.append(y0)

    if not men_positions:
        raise ValueError("Men heading not found in table area.")

    if not women_positions:
        raise ValueError("Women heading not found in table area.")

    men_y = min(men_positions)
    women_y = min(women_positions)

    if women_y <= men_y:
        raise ValueError("Invalid Men/Women section positions.")

    return {
        "Men": {
            "heading_y": men_y,
            "start_y": men_y + 15,
            "end_y": women_y - 5
        },
        "Women": {
            "heading_y": women_y,
            "start_y": women_y + 10,
            "end_y": 400
        }
    }


attire_sections = identify_attire_sections(page_16)

print("\nIDENTIFIED TABLE SECTIONS")
print("=" * 60)

for category, bounds in attire_sections.items():

    print("\n", category)

    for key, value in bounds.items():
        print(key, ":", round(value, 2))


# ------------------------------------------------
# STEP 9: GROUP WORDS INTO TABLE LINES
# ------------------------------------------------

def extract_table_lines(page, y_min, y_max):

    words = page.get_text("words", sort=True)

    selected_words = []

    for word in words:

        x0, y0, x1, y1, text = word[:5]

        if y_min <= y0 <= y_max:

            selected_words.append({
                "x": x0,
                "y": y0,
                "text": text
            })

    selected_words.sort(
        key=lambda item: (item["y"], item["x"])
    )

    grouped_lines = []

    for word in selected_words:

        if (
            not grouped_lines
            or abs(word["y"] - grouped_lines[-1]["y"]) > 2.5
        ):

            grouped_lines.append({
                "y": word["y"],
                "words": [word]
            })

        else:

            grouped_lines[-1]["words"].append(word)

            grouped_lines[-1]["y"] = sum(
                item["y"]
                for item in grouped_lines[-1]["words"]
            ) / len(grouped_lines[-1]["words"])

    structured_lines = []

    for line in grouped_lines:

        left = []
        middle = []
        right = []

        for word in sorted(
            line["words"],
            key=lambda item: item["x"]
        ):

            x = word["x"]
            text = word["text"]

            if x < 270:
                left.append(text)

            elif 280 <= x <= 305:
                middle.append(text)

            elif x > 305:
                right.append(text)

        structured_lines.append({
            "y": round(line["y"], 2),
            "left": " ".join(left).strip(),
            "number": " ".join(middle).strip(),
            "right": " ".join(right).strip()
        })

    return structured_lines


# ------------------------------------------------
# STEP 10: RECONSTRUCT THE DRESS-CODE TABLE
# ------------------------------------------------

def reconstruct_attire_table(page, page_number):

    sections = identify_attire_sections(page)

    all_records = []

    for category, bounds in sections.items():

        lines = extract_table_lines(
            page,
            bounds["start_y"],
            bounds["end_y"]
        )

        anchors = []

        for line_index, line in enumerate(lines):

            middle_number = line["number"].strip()

            # Number found in the middle column
            if middle_number in ["1", "2", "3", "4", "5", "6"]:

                anchors.append({
                    "line_index": line_index,
                    "number": middle_number,
                    "y": line["y"]
                })

            # Some women's numbering is merged into the left text
            else:

                match = re.match(
                    r"^([1-6])(?=[A-Za-z])",
                    line["left"]
                )

                if category == "Women" and match:

                    anchors.append({
                        "line_index": line_index,
                        "number": match.group(1),
                        "y": line["y"]
                    })

        # Remove duplicate anchors at nearly identical positions
        unique_anchors = []

        for anchor in anchors:

            duplicate = any(
                abs(anchor["y"] - existing["y"]) <= 2.5
                for existing in unique_anchors
            )

            if not duplicate:
                unique_anchors.append(anchor)

        unique_anchors.sort(key=lambda item: item["y"])

        # Build one record for every numbered row
        for anchor_index, anchor in enumerate(unique_anchors):

            start_index = anchor["line_index"]

            if anchor_index + 1 < len(unique_anchors):
                end_index = unique_anchors[
                    anchor_index + 1
                ]["line_index"]
            else:
                end_index = len(lines)

            row_lines = lines[start_index:end_index]

            acceptable_parts = []
            unacceptable_parts = []

            for line in row_lines:

                left_text = line["left"]
                right_text = line["right"]

                # Remove merged numbering from women's left text
                if (
                    category == "Women"
                    and left_text
                ):

                    left_text = re.sub(
                        r"^[1-6](?=[A-Za-z])",
                        "",
                        left_text,
                        count=1
                    )

                if left_text:
                    acceptable_parts.append(left_text)

                if right_text:
                    unacceptable_parts.append(right_text)

            acceptable_attire = " ".join(
                acceptable_parts
            ).strip()

            unacceptable_attire = " ".join(
                unacceptable_parts
            ).strip()

            all_records.append({

                "document_name": pdf_filename,
                "page_number": page_number,
                "category": category,
                "row_number": int(anchor["number"]),
                "acceptable_attire": acceptable_attire,
                "unacceptable_number": int(anchor["number"]),
                "unacceptable_attire": unacceptable_attire

            })

    return all_records


# ------------------------------------------------
# STEP 11: RUN RECONSTRUCTION
# ------------------------------------------------

attire_records = reconstruct_attire_table(
    page=page_16,
    page_number=16
)

attire_df = pd.DataFrame(attire_records)

print("\nRECONSTRUCTION RESULTS")
print("=" * 60)
print("Records reconstructed:", len(attire_records))
print("DataFrame columns:", attire_df.columns.tolist())


# ------------------------------------------------
# STEP 12: DISPLAY RECONSTRUCTED TABLE
# ------------------------------------------------

if not attire_df.empty:

    display(
        attire_df[
            [
                "page_number",
                "category",
                "row_number",
                "acceptable_attire",
                "unacceptable_number",
                "unacceptable_attire"
            ]
        ]
    )

else:

    print("No records were reconstructed.")


# ------------------------------------------------
# STEP 13: FINAL VALIDATION
# ------------------------------------------------

print("\nFINAL PROCESSING SUMMARY")
print("=" * 60)

print("PDF:", pdf_filename)
print("Total pages:", len(documents))
print("Tables detected:", len(extracted_tables))
print("Empty pages:", len(empty_pages))
print("Original text characters:", original_total)
print("Cleaned text characters:", cleaned_total)
print("Page 16 reconstructed rows:", len(attire_records))

if not attire_df.empty:

    print("\nRows by category:")

    print(
        attire_df.groupby("category").size()
    )

    print("\nMissing acceptable attire values:")
    print(
        attire_df["acceptable_attire"].eq("").sum()
    )

    print("\nMissing unacceptable attire values:")
    print(
        attire_df["unacceptable_attire"].eq("").sum()
    )

print("\nDOCUMENT PROCESSING STAGE COMPLETED.")
```

```python

# ================================================================
# HR POLICY AI PROJECT
# STAGE 2: DOCUMENT STRUCTURE AND INTELLIGENT CHUNKING
# ================================================================
# 1. Identify sections and chapters
# 2. Organize extracted text into paragraphs
# 3. Create manageable text chunks
# 4. Preserve page numbers and document references
# 5. Add the reconstructed dress-code table as a separate chunk
# 6. Validate the resulting chunks
# ================================================================

import re
import pandas as pd


# ------------------------------------------------
# STEP 1: CONFIGURATION
# ------------------------------------------------

CHUNK_SIZE = 1600
CHUNK_OVERLAP = 250

chunks = []

current_section = None
current_chapter = None


# ------------------------------------------------
# STEP 2: IDENTIFY DOCUMENT HEADINGS
# ------------------------------------------------

def identify_heading(line):

    line = line.strip()

    if not line:
        return None

    # Section heading
    if re.match(
        r"^SECTION\s+[IVX0-9]+",
        line,
        re.IGNORECASE
    ):
        return "section"

    # Chapter heading
    if re.match(
        r"^CHAPTER\s+\d+",
        line,
        re.IGNORECASE
    ):
        return "chapter"

    # Numbered subsection heading, e.g. 1.1 About Us
    if re.match(
        r"^\d+(?:\.\d+)+\.?\s+[A-Z]",
        line
    ):
        return "subsection"

    return None


# ------------------------------------------------
# STEP 3: ORGANIZE PAGE TEXT INTO PARAGRAPHS
# ------------------------------------------------

def organize_paragraphs(text):

    lines = text.splitlines()

    paragraphs = []
    current_paragraph = []

    for line in lines:

        line = line.strip()

        if not line:

            if current_paragraph:
                paragraphs.append(
                    " ".join(current_paragraph)
                )
                current_paragraph = []

            continue

        heading_type = identify_heading(line)

        is_bullet = bool(
            re.match(
                r"^(?:[-•▪]|[ivxlcdm]+\)|\d+[.)])\s*",
                line,
                re.IGNORECASE
            )
        )

        # Keep headings and list items separate
        if heading_type or is_bullet:

            if current_paragraph:
                paragraphs.append(
                    " ".join(current_paragraph)
                )
                current_paragraph = []

            paragraphs.append(line)

        else:

            current_paragraph.append(line)

    if current_paragraph:
        paragraphs.append(
            " ".join(current_paragraph)
        )

    return [
        paragraph.strip()
        for paragraph in paragraphs
        if paragraph.strip()
    ]


# ------------------------------------------------
# STEP 4: SPLIT LONG PARAGRAPHS
# ------------------------------------------------

def split_long_text(text, max_length):

    if len(text) <= max_length:
        return [text]

    sentences = re.split(
        r"(?<=[.!?])\s+",
        text
    )

    pieces = []
    current_piece = ""

    for sentence in sentences:

        if not current_piece:

            current_piece = sentence

        elif len(current_piece) + len(sentence) + 1 <= max_length:

            current_piece += " " + sentence

        else:

            pieces.append(current_piece.strip())
            current_piece = sentence

    if current_piece:
        pieces.append(current_piece.strip())

    # Handle unusually long individual sentences
    final_pieces = []

    for piece in pieces:

        while len(piece) > max_length:

            split_position = piece.rfind(
                " ",
                0,
                max_length
            )

            if split_position == -1:
                split_position = max_length

            final_pieces.append(
                piece[:split_position].strip()
            )

            piece = piece[split_position:].strip()

        if piece:
            final_pieces.append(piece)

    return final_pieces


# ------------------------------------------------
# STEP 5: CREATE CHUNKS WHILE PRESERVING CONTEXT
# ------------------------------------------------

def create_page_chunks(
    paragraphs,
    page_number,
    document_name,
    chapter,
    section
):

    page_chunks = []

    current_text = ""

    def save_chunk(text):

        if not text.strip():
            return

        page_chunks.append({

            "document_name": document_name,
            "page_number": page_number,
            "chapter": chapter,
            "section": section,
            "chunk_text": text.strip(),
            "character_count": len(text.strip()),
            "content_type": "page_text"

        })

    for paragraph in paragraphs:

        pieces = split_long_text(
            paragraph,
            CHUNK_SIZE
        )

        for piece in pieces:

            if not current_text:

                current_text = piece

            elif (
                len(current_text) + len(piece) + 2
                <= CHUNK_SIZE
            ):

                current_text += "\n\n" + piece

            else:

                previous_text = current_text

                save_chunk(previous_text)

                # Preserve a small amount of previous context
                overlap_text = previous_text[
                    -CHUNK_OVERLAP:
                ].strip()

                if len(overlap_text) >= CHUNK_OVERLAP * 0.8:

                    current_text = (
                        overlap_text + "\n\n" + piece
                    )

                else:

                    current_text = piece

    if current_text:
        save_chunk(current_text)

    return page_chunks


# ------------------------------------------------
# STEP 6: PROCESS ALL PAGES
# ------------------------------------------------

chunks = []

current_section = None
current_chapter = None

for document in cleaned_documents:

    page_number = document["page_number"]
    page_text = document["cleaned_text"]

    paragraphs = organize_paragraphs(page_text)

    # Update chapter and section context
    for paragraph in paragraphs:

        heading_type = identify_heading(paragraph)

        if heading_type == "section":
            current_section = paragraph

        elif heading_type == "chapter":
            current_chapter = paragraph

    # Create chunks for this page
    page_chunks = create_page_chunks(
        paragraphs=paragraphs,
        page_number=page_number,
        document_name=document["document_name"],
        chapter=current_chapter,
        section=current_section
    )

    chunks.extend(page_chunks)


# ------------------------------------------------
# STEP 7: ADD RECONSTRUCTED DRESS-CODE TABLE
# ------------------------------------------------

if not attire_df.empty:

    table_lines = []

    table_lines.append(
        "HR POLICY MANUAL - DRESS CODE"
    )

    table_lines.append(
        "Source: Page 16"
    )

    for _, row in attire_df.iterrows():

        table_lines.append(
            f"Category: {row['category']}"
        )

        table_lines.append(
            f"Acceptable attire: {row['acceptable_attire']}"
        )

        table_lines.append(
            f"Unacceptable attire: {row['unacceptable_attire']}"
        )

        table_lines.append("")

    structured_table_text = "\n".join(table_lines)

    chunks.append({

        "document_name": pdf_filename,
        "page_number": 16,
        "chapter": current_chapter,
        "section": current_section,
        "chunk_text": structured_table_text,
        "character_count": len(structured_table_text),
        "content_type": "structured_table"

    })


# ------------------------------------------------
# STEP 8: CREATE CHUNK DATAFRAME
# ------------------------------------------------

chunks_df = pd.DataFrame(chunks)

chunks_df.insert(
    0,
    "chunk_id",
    [
        f"HR_CHUNK_{i+1:04d}"
        for i in range(len(chunks_df))
    ]
)


# ------------------------------------------------
# STEP 9: VALIDATE CHUNKS
# ------------------------------------------------

print("\nCHUNKING RESULTS")
print("=" * 65)

print("Total pages processed:", len(cleaned_documents))
print("Total chunks created:", len(chunks_df))

print(
    "Average chunk size:",
    round(chunks_df["character_count"].mean(), 2)
)

print(
    "Smallest chunk:",
    chunks_df["character_count"].min()
)

print(
    "Largest chunk:",
    chunks_df["character_count"].max()
)

print(
    "Chunks exceeding configured size:",
    (chunks_df["character_count"] > CHUNK_SIZE).sum()
)

print(
    "Chunks without page numbers:",
    chunks_df["page_number"].isna().sum()
)

print(
    "Structured table chunks:",
    (chunks_df["content_type"] == "structured_table").sum()
)


# ------------------------------------------------
# STEP 10: INSPECT CHUNK DISTRIBUTION
# ------------------------------------------------

print("\nCHUNKS BY CONTENT TYPE")
print("=" * 65)

display(
    chunks_df["content_type"].value_counts().to_frame(
        "Number of chunks"
    )
)

print("\nCHUNKS BY PAGE - FIRST 15 PAGES")
print("=" * 65)

display(
    chunks_df.groupby("page_number")
    .size()
    .reset_index(name="number_of_chunks")
    .head(15)
)


# ------------------------------------------------
# STEP 11: INSPECT SAMPLE CHUNKS
# ------------------------------------------------

print("\nSAMPLE CHUNK 1")
print("=" * 65)

print(chunks_df.iloc[0]["chunk_text"])

print("\nSAMPLE CHUNK 2")
print("=" * 65)

if len(chunks_df) > 1:
    print(chunks_df.iloc[1]["chunk_text"])


# ------------------------------------------------
# STEP 12: INSPECT STRUCTURED TABLE CHUNK
# ------------------------------------------------

table_chunks = chunks_df[
    chunks_df["content_type"] == "structured_table"
]

if not table_chunks.empty:

    print("\nSTRUCTURED DRESS-CODE TABLE CHUNK")
    print("=" * 65)

    print(table_chunks.iloc[0]["chunk_text"])


# ------------------------------------------------
# STEP 13: FINAL VALIDATION
# ------------------------------------------------

print("\nFINAL STAGE 2 SUMMARY")
print("=" * 65)

print("Document:", pdf_filename)
print("Pages:", len(cleaned_documents))
print("Chunks:", len(chunks_df))
print("Chunk size:", CHUNK_SIZE)
print("Overlap target:", CHUNK_OVERLAP)

print(
    "All chunks have text:",
    chunks_df["chunk_text"].notna().all()
)

print(
    "All chunks have page references:",
    chunks_df["page_number"].notna().all()
)

print(
    "All chunks have unique IDs:",
    chunks_df["chunk_id"].is_unique
)

print("\nDOCUMENT CHUNKING COMPLETED.")
```

```python

# ============================================================
# STAGE 2B: PROCESS ALL DETECTED TABLES
# ============================================================

import pandas as pd
import re

print("=" * 65)
print("PROCESSING ALL HR POLICY TABLES")
print("=" * 65)

# ------------------------------------------------------------
# 1. CHECK THAT THE REQUIRED DATA EXISTS
# ------------------------------------------------------------

if "extracted_tables" not in globals():
    raise ValueError("extracted_tables was not found. Run Stage 1 first.")

if "chunks" not in globals():
    raise ValueError("chunks was not found. Run the original Stage 2 first.")

print("Detected tables:", len(extracted_tables))
print("Existing chunks:", len(chunks))

# We expect 35 tables based on Stage 1.
if len(extracted_tables) == 0:
    raise ValueError("No extracted tables were found.")

# ------------------------------------------------------------
# 2. KEEP NORMAL TEXT CHUNKS
# Remove the previous page 16 table chunk to avoid duplication.
# ------------------------------------------------------------

text_chunk_records = []

for chunk in chunks:
    if chunk.get("content_type") != "structured_table":
        text_chunk_records.append(dict(chunk))

print("Normal text chunks retained:", len(text_chunk_records))


# ------------------------------------------------------------
# 3. CLEAN TABLE CELL VALUES
# ------------------------------------------------------------

def clean_table_cell(value):

    if value is None:
        return ""

    try:
        if pd.isna(value):
            return ""
    except (TypeError, ValueError):
        pass

    value = str(value)
    value = value.replace("\n", " ")
    value = re.sub(r"\s+", " ", value)

    return value.strip()


# ------------------------------------------------------------
# 4. CONVERT A TABLE INTO READABLE ROW TEXT
# ------------------------------------------------------------

def convert_table_rows_to_text(rows):

    if not rows:
        return [], 0, 0, 0

    cleaned_rows = []

    for row in rows:

        if row is None:
            row = []

        cleaned_row = [
            clean_table_cell(cell)
            for cell in row
        ]

        cleaned_rows.append(cleaned_row)

    max_columns = max(
        (len(row) for row in cleaned_rows),
        default=0
    )

    # Make all rows have the same number of columns.
    normalized_rows = []

    for row in cleaned_rows:

        normalized_row = row + [""] * (
            max_columns - len(row)
        )

        normalized_rows.append(normalized_row)

    # Count empty cells for extraction-quality checks.
    total_cells = len(normalized_rows) * max_columns

    empty_cells = sum(
        1
        for row in normalized_rows
        for cell in row
        if not cell
    )

    return (
        normalized_rows,
        len(cleaned_rows),
        max_columns,
        empty_cells
    )


# ------------------------------------------------------------
# 5. GROUP ROWS INTO CHUNKS
# Keep complete rows together wherever possible.
# ------------------------------------------------------------

TABLE_CHUNK_LIMIT = 1500


def create_table_chunks(
    page_number,
    table_number,
    row_lines,
    extraction_status
):

    result = []
    current_lines = []

    source_header = (
        f"HR POLICY MANUAL TABLE\n"
        f"Source page: {page_number}\n"
        f"Table number: {table_number}\n"
        f"Extraction status: {extraction_status}\n"
    )

    for line in row_lines:

        proposed_text = (
            source_header
            + "\n"
            + "\n".join(current_lines + [line])
        )

        if (
            len(proposed_text) > TABLE_CHUNK_LIMIT
            and current_lines
        ):

            result.append(
                source_header
                + "\n"
                + "\n".join(current_lines)
            )

            current_lines = []

        # Handle an unusually long individual row.
        if len(source_header) + len(line) > TABLE_CHUNK_LIMIT:

            if current_lines:
                result.append(
                    source_header
                    + "\n"
                    + "\n".join(current_lines)
                )
                current_lines = []

            available_length = (
                TABLE_CHUNK_LIMIT - len(source_header) - 30
            )

            pieces = [
                line[i:i + available_length]
                for i in range(0, len(line), available_length)
            ]

            for piece_number, piece in enumerate(pieces, start=1):

                result.append(
                    source_header
                    + f"\nContinuation of row: {piece_number}\n"
                    + piece
                )

        else:
            current_lines.append(line)

    if current_lines:

        result.append(
            source_header
            + "\n"
            + "\n".join(current_lines)
        )

    return result


# ------------------------------------------------------------
# 6. PROCESS ALL 35 TABLES
# ------------------------------------------------------------

table_chunk_records = []
table_quality_records = []

for table in extracted_tables:

    page_number = int(table["page_number"])
    table_number = int(table["table_number"])

    # --------------------------------------------------------
    # SPECIAL CASE: PAGE 16 DRESS-CODE TABLE
    # Use the corrected reconstruction from the earlier step.
    # --------------------------------------------------------

    if page_number == 16:

        if "attire_df" not in globals():
            raise ValueError(
                "The corrected attire_df was not found. "
                "Run the page 16 reconstruction cell first."
            )

        required_columns = [
            "category",
            "acceptable_attire",
            "unacceptable_attire"
        ]

        missing_columns = [
            column
            for column in required_columns
            if column not in attire_df.columns
        ]

        if missing_columns:
            raise ValueError(
                f"Missing columns in attire_df: {missing_columns}"
            )

        row_lines = []

        for index, row in attire_df.iterrows():

            category = clean_table_cell(row["category"])
            acceptable = clean_table_cell(
                row["acceptable_attire"]
            )
            unacceptable = clean_table_cell(
                row["unacceptable_attire"]
            )

            row_lines.append(
                f"Category: {category}\n"
                f"Acceptable attire: {acceptable}\n"
                f"Unacceptable attire: {unacceptable}"
            )

        extraction_status = "corrected_manual_reconstruction"

        table_quality_records.append({
            "page_number": page_number,
            "table_number": table_number,
            "row_count": len(attire_df),
            "column_count": 3,
            "empty_cells": 0,
            "quality_status": extraction_status
        })

    # --------------------------------------------------------
    # ALL OTHER TABLES: USE PYMUPDF EXTRACTED ROWS
    # --------------------------------------------------------

    else:

        raw_rows = table.get("rows", [])

        (
            normalized_rows,
            row_count,
            column_count,
            empty_cells
        ) = convert_table_rows_to_text(raw_rows)

        row_lines = []

        for row_index, row in enumerate(
            normalized_rows,
            start=1
        ):

            cell_descriptions = []

            for column_index, cell in enumerate(
                row,
                start=1
            ):

                cell_value = cell if cell else "[EMPTY]"

                cell_descriptions.append(
                    f"Column {column_index}: {cell_value}"
                )

            row_lines.append(
                f"Row {row_index}: "
                + " | ".join(cell_descriptions)
            )

        # Flag possible extraction problems.
        original_widths = [
            len(row) if row is not None else 0
            for row in raw_rows
        ]

        inconsistent_widths = (
            len(set(original_widths)) > 1
            if original_widths
            else False
        )

        empty_ratio = (
            empty_cells / (row_count * column_count)
            if row_count and column_count
            else 1
        )

        if (
            row_count == 0
            or column_count == 0
            or empty_ratio > 0.25
            or inconsistent_widths
        ):
            extraction_status = "review_required"
        else:
            extraction_status = "extracted_needs_visual_check"

        table_quality_records.append({
            "page_number": page_number,
            "table_number": table_number,
            "row_count": row_count,
            "column_count": column_count,
            "empty_cells": empty_cells,
            "quality_status": extraction_status
        })

    # --------------------------------------------------------
    # CREATE ONE OR MORE CHUNKS FOR THIS TABLE
    # --------------------------------------------------------

    table_text_chunks = create_table_chunks(
        page_number=page_number,
        table_number=table_number,
        row_lines=row_lines,
        extraction_status=extraction_status
    )

    for part_number, table_text in enumerate(
        table_text_chunks,
        start=1
    ):

        table_chunk_records.append({
            "chunk_id": (
                f"TABLE_P{page_number}_"
                f"T{table_number}_C{part_number}"
            ),
            "document_name": table.get(
                "document_name",
                pdf_filename
            ),
            "page_number": page_number,
            "chapter": "",
            "section": "",
            "table_number": table_number,
            "chunk_text": table_text,
            "character_count": len(table_text),
            "content_type": "structured_table",
            "extraction_status": extraction_status,
            "row_count": (
                len(attire_df)
                if page_number == 16
                else table.get("row_count", 0)
            ),
            "column_count": (
                3
                if page_number == 16
                else table.get("column_count", 0)
            )
        })


# ------------------------------------------------------------
# 7. COMBINE TEXT AND TABLE CHUNKS
# ------------------------------------------------------------

chunks = text_chunk_records + table_chunk_records

chunks_df = pd.DataFrame(chunks)

# Ensure every chunk has a unique ID.
if chunks_df["chunk_id"].duplicated().any():
    raise ValueError("Duplicate chunk IDs were detected.")

# ------------------------------------------------------------
# 8. CREATE TABLE QUALITY REPORT
# ------------------------------------------------------------

table_quality_df = pd.DataFrame(table_quality_records)

print("\n" + "=" * 65)
print("TABLE PROCESSING RESULTS")
print("=" * 65)

print("Tables detected:", len(extracted_tables))
print("Tables processed:", len(table_quality_df))
print("Table chunks created:", len(table_chunk_records))
print("Normal text chunks retained:", len(text_chunk_records))
print("Total chunks now:", len(chunks_df))

print("\nTABLE EXTRACTION QUALITY")
print("=" * 65)

print(
    table_quality_df["quality_status"]
    .value_counts()
    .to_string()
)

print("\nTABLE-BY-TABLE REVIEW")
print("=" * 65)

display(
    table_quality_df.sort_values(
        ["page_number", "table_number"]
    ).reset_index(drop=True)
)

# ------------------------------------------------------------
# 9. FINAL CHUNK VALIDATION
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("FINAL STAGE 2 VALIDATION")
print("=" * 65)

print("Total pages processed:", len(cleaned_documents))
print("Total chunks:", len(chunks_df))

print(
    "Normal text chunks:",
    len(chunks_df[chunks_df["content_type"] == "page_text"])
)

print(
    "Structured table chunks:",
    len(chunks_df[
        chunks_df["content_type"] == "structured_table"
    ])
)

print(
    "Chunks without text:",
    chunks_df["chunk_text"].isna().sum()
    + (chunks_df["chunk_text"].str.strip() == "").sum()
)

print(
    "Chunks without page numbers:",
    chunks_df["page_number"].isna().sum()
)

print(
    "Duplicate chunk IDs:",
    chunks_df["chunk_id"].duplicated().sum()
)

print(
    "Largest chunk:",
    chunks_df["character_count"].max()
)

print(
    "Chunks exceeding 1600 characters:",
    (chunks_df["character_count"] > 1600).sum()
)

print("\nSTAGE 2B COMPLETED.")
```

```python

# ============================================================
# STAGE 2C: RECONSTRUCT TABLES ON PAGES 7 AND 18
# ============================================================

def clean_table_text(value):
    if value is None:
        return ""

    value = str(value)
    value = value.replace("\n", " ")
    value = re.sub(r"\s+", " ", value)
    value = re.sub(r"\s+([,.;])", r"\1", value)

    return value.strip(" ,")


# ------------------------------------------------------------
# PAGE 7 - TABLE 2
# Logical columns: Skills Level, Grade Description,
# Grade, Qualifications
# ------------------------------------------------------------

page7_table = next(
    t for t in extracted_tables
    if t["page_number"] == 7 and t["table_number"] == 2
)

page7_rows = page7_table["rows"]

page7_records = []

for row in page7_rows[1:]:

    skills_level = clean_table_text(row[0])
    grade_description = clean_table_text(row[3])
    grade = clean_table_text(row[6])
    qualification = clean_table_text(row[10])

    if not qualification:
        qualification = (
            "[Not separately extracted - verify against original PDF]"
        )

    page7_records.append({
        "Skills Level": skills_level,
        "Grade Description": grade_description,
        "Grade": grade,
        "Qualifications": qualification
    })


# ------------------------------------------------------------
# PAGE 18 - TABLE 1
# Logical columns: Conflict Type, Examples
# ------------------------------------------------------------

page18_table = next(
    t for t in extracted_tables
    if t["page_number"] == 18 and t["table_number"] == 1
)

page18_rows = page18_table["rows"]

page18_records = []

for row in page18_rows[1:]:

    conflict_type = clean_table_text(row[0])
    examples = clean_table_text(row[3])

    page18_records.append({
        "Conflict Type": conflict_type,
        "Examples": examples
    })


# ------------------------------------------------------------
# DISPLAY RECONSTRUCTED TABLES
# ------------------------------------------------------------

page7_df = pd.DataFrame(page7_records)
page18_df = pd.DataFrame(page18_records)

print("PAGE 7 - TABLE 2: STAFF GRADING")
display(page7_df)

print("\nPAGE 18 - TABLE 1: CONFLICT OF INTEREST")
display(page18_df)


# ------------------------------------------------------------
# CONVERT RECONSTRUCTED TABLES INTO TEXT CHUNKS
# ------------------------------------------------------------

def records_to_row_lines(records):

    lines = []

    for index, record in enumerate(records, start=1):

        line = f"Row {index}: " + " | ".join(
            f"{key}: {value}"
            for key, value in record.items()
        )

        lines.append(line)

    return lines


corrected_table_chunks = []

tables_to_reconstruct = [
    {
        "page": 7,
        "table": 2,
        "title": "Staff Grading",
        "records": page7_records
    },
    {
        "page": 18,
        "table": 1,
        "title": "Conflict of Interest",
        "records": page18_records
    }
]

for item in tables_to_reconstruct:

    lines = records_to_row_lines(item["records"])

    header = (
        f"HR POLICY MANUAL - {item['title']}\n"
        f"Source page: {item['page']}\n"
        f"Table number: {item['table']}\n"
        f"Extraction status: corrected logical columns\n"
    )

    table_texts = create_table_chunks(
        page_number=item["page"],
        table_number=item["table"],
        row_lines=lines,
        extraction_status="corrected_logical_columns"
    )

    for part_number, text in enumerate(table_texts, start=1):

        corrected_table_chunks.append({
            "chunk_id": (
                f"CORRECTED_P{item['page']}_"
                f"T{item['table']}_C{part_number}"
            ),
            "document_name": pdf_filename,
            "page_number": item["page"],
            "chapter": "",
            "section": "",
            "table_number": item["table"],
            "chunk_text": header + "\n" + text,
            "character_count": len(header + "\n" + text),
            "content_type": "structured_table",
            "extraction_status": "corrected_logical_columns",
            "row_count": len(item["records"]),
            "column_count": (
                4 if item["page"] == 7 else 2
            )
        })


# ------------------------------------------------------------
# REPLACE THE OLD VERSIONS IN THE MAIN DATASET
# ------------------------------------------------------------

replacement_keys = {(7, 2), (18, 1)}

keep_mask = ~(
    (chunks_df["content_type"] == "structured_table")
    & chunks_df.apply(
        lambda r: (
            int(r["page_number"]),
            int(r["table_number"])
        ) in replacement_keys,
        axis=1
    )
)

chunks_df = chunks_df.loc[keep_mask].copy()

corrected_df = pd.DataFrame(corrected_table_chunks)

chunks_df = pd.concat(
    [chunks_df, corrected_df],
    ignore_index=True
)

# Reassign unique IDs across the complete dataset.
chunks_df["chunk_id"] = [
    f"HR_CHUNK_{i:06d}"
    for i in range(1, len(chunks_df) + 1)
]

chunks = chunks_df.to_dict("records")

# Update the quality report.
for page, table_number in [(7, 2), (18, 1)]:

    mask = (
        (table_quality_df["page_number"] == page)
        & (table_quality_df["table_number"] == table_number)
    )

    table_quality_df.loc[mask, "quality_status"] = (
        "corrected_logical_columns_needs_visual_check"
    )

print("\nRECONSTRUCTION SUMMARY")
print("=" * 55)
print("Corrected tables: 2")
print("Total chunks:", len(chunks_df))
print("Duplicate IDs:", chunks_df["chunk_id"].duplicated().sum())

print("\nUpdated quality report:")
display(table_quality_df)
```

```python
# ============================================================
# UPDATE PAGE 21 AND PAGE 22 STORED CHUNKS
# ============================================================

def dataframe_to_table_text(title, page_number, table_number,
                            dataframe, continuation=None):

    text = (
        f"HR POLICY MANUAL - {title}\n"
        f"Source page: {page_number}\n"
        f"Table number: {table_number}\n"
        f"Extraction status: corrected_manual_verified\n"
    )

    if continuation:
        text += f"Table relationship: {continuation}\n"

    text += "\nHR POLICY MANUAL TABLE\n"

    for index, row in dataframe.iterrows():
        row_values = [
            f"{column}: {row[column]}"
            for column in dataframe.columns
        ]

        text += f"\nRow {index + 1}: " + " | ".join(row_values)

    return text


# Define the four replacements
table_updates = [
    {
        "page": 21,
        "table": 1,
        "df": page21_table1_df,
        "title": "Misconduct Categories",
        "continuation": "Continuation of page 20, Table 1"
    },
    {
        "page": 21,
        "table": 2,
        "df": page21_table2_df,
        "title": "Reporting Channels",
        "continuation": None
    },
    {
        "page": 21,
        "table": 3,
        "df": page21_table3_df,
        "title": "Reporting Responsibilities",
        "continuation": "Continues on page 22, Table 1"
    },
    {
        "page": 22,
        "table": 1,
        "df": page22_table1_df,
        "title": "Reporting Responsibilities",
        "continuation": "Continuation of page 21, Table 3"
    }
]


# Update each existing chunk
for item in table_updates:

    page = item["page"]
    table = item["table"]
    dataframe = item["df"]

    mask = (
        (pd.to_numeric(chunks_df["page_number"], errors="coerce") == page) &
        (pd.to_numeric(chunks_df["table_number"], errors="coerce") == table) &
        (chunks_df["content_type"] == "structured_table")
    )

    assert mask.sum() == 1, (
        f"Expected exactly one chunk for page {page}, table {table}. "
        f"Found: {mask.sum()}"
    )

    idx = chunks_df.index[mask][0]

    corrected_text = dataframe_to_table_text(
        title=item["title"],
        page_number=page,
        table_number=table,
        dataframe=dataframe,
        continuation=item["continuation"]
    )

    chunks_df.at[idx, "chunk_text"] = corrected_text
    chunks_df.at[idx, "character_count"] = len(corrected_text)
    chunks_df.at[idx, "extraction_status"] = "corrected_manual_verified"
    chunks_df.at[idx, "row_count"] = len(dataframe)
    chunks_df.at[idx, "column_count"] = len(dataframe.columns)


# Synchronize the original chunks list
chunks = chunks_df.to_dict("records")

print("PAGE 21 AND PAGE 22 TABLE UPDATES COMPLETED")
print("=" * 60)
```

```python
# ============================================================
# PAGE 82 — SAVE BOTH VERIFIED TABLES
# ============================================================

def update_table_chunk(
    page_number,
    table_number,
    dataframe,
    extra_text=None
):
    mask = (
        (pd.to_numeric(chunks_df["page_number"], errors="coerce") == page_number) &
        (pd.to_numeric(chunks_df["table_number"], errors="coerce") == table_number) &
        (chunks_df["content_type"] == "structured_table")
    )

    assert mask.sum() == 1, (
        f"Expected one chunk for page {page_number}, "
        f"table {table_number}; found {mask.sum()}"
    )

    idx = chunks_df.index[mask][0]

    text_content = dataframe_to_table_text(
        dataframe,
        page_number=page_number,
        table_number=table_number
    )

    if extra_text:
        text_content += "\n\n" + extra_text

    chunks_df.at[idx, "chunk_text"] = text_content
    chunks_df.at[idx, "character_count"] = len(text_content)
    chunks_df.at[idx, "row_count"] = len(dataframe)
    chunks_df.at[idx, "column_count"] = len(dataframe.columns)
    chunks_df.at[idx, "extraction_status"] = "corrected_manual_verified"

    return chunks_df.at[idx, "chunk_id"]


# Update Table 1
table1_id = update_table_chunk(
    page_number=82,
    table_number=1,
    dataframe=page82_table1_df
)

# Update Table 2 and preserve its footnote
table2_id = update_table_chunk(
    page_number=82,
    table_number=2,
    dataframe=page82_table2_df,
    extra_text=(
        "IMPORTANT FOOTNOTE: Corporate Office Car monthly figure "
        "is excluded from the monthly pool total; entitlement is "
        "billed on a quarterly basis."
    )
)

# Synchronize records
chunks = chunks_df.to_dict("records")

print("PAGE 82 UPDATED SUCCESSFULLY")
print("Table 1 chunk:", table1_id)
print("Table 2 chunk:", table2_id)
```

```python

# ============================================================
# CREATE EMBEDDING-READY CHUNKS
# ============================================================

import pandas as pd
import math

MAX_CONTENT_TOKENS = 200
TOKEN_OVERLAP = 30

tokenizer = embedding_model.tokenizer

# Confirm that the tokenizer supports character offsets
assert tokenizer.is_fast, "A fast tokenizer is required for this splitting method."

embedding_chunks = []

for _, row in chunks_df.iterrows():

    original_text = str(row["chunk_text"])

    # Tokenize without special tokens so we can track exact text positions
    encoded = tokenizer(
        original_text,
        add_special_tokens=False,
        return_offsets_mapping=True,
        truncation=False
    )

    offsets = encoded["offset_mapping"]
    total_tokens = len(offsets)

    # Preserve short chunks without splitting
    if total_tokens <= MAX_CONTENT_TOKENS:

        new_row = row.to_dict()
        new_row["original_chunk_id"] = row["chunk_id"]
        new_row["embedding_chunk_id"] = row["chunk_id"] + "_PART_001"
        new_row["part_number"] = 1
        new_row["token_count"] = len(
            tokenizer.encode(original_text, add_special_tokens=True)
        )

        embedding_chunks.append(new_row)

    else:

        start_token = 0
        part_number = 1

        while start_token < total_tokens:

            end_token = min(
                start_token + MAX_CONTENT_TOKENS,
                total_tokens
            )

            # Convert token positions into character positions
            start_char = offsets[start_token][0]
            end_char = offsets[end_token - 1][1]

            piece = original_text[start_char:end_char].strip()

            if piece:

                new_row = row.to_dict()

                new_row["original_chunk_id"] = row["chunk_id"]
                new_row["embedding_chunk_id"] = (
                    f"{row['chunk_id']}_PART_{part_number:03d}"
                )
                new_row["part_number"] = part_number
                new_row["chunk_text"] = piece
                new_row["character_count"] = len(piece)

                new_row["token_count"] = len(
                    tokenizer.encode(piece, add_special_tokens=True)
                )

                embedding_chunks.append(new_row)

            # Move forward, retaining overlapping context
            if end_token == total_tokens:
                break

            start_token = end_token - TOKEN_OVERLAP
            part_number += 1


# Create a separate dataframe
embedding_chunks_df = pd.DataFrame(embedding_chunks)

# ============================================================
# VALIDATION
# ============================================================

print("EMBEDDING CHUNK GENERATION COMPLETE")
print("=" * 60)

print("Original chunks:", len(chunks_df))
print("New embedding chunks:", len(embedding_chunks_df))

print("\nTOKEN VALIDATION")
print("=" * 60)

print("Model maximum:", embedding_model.max_seq_length)
print("Largest embedding chunk:", embedding_chunks_df["token_count"].max())

over_limit = embedding_chunks_df[
    embedding_chunks_df["token_count"] > embedding_model.max_seq_length
]

print("Chunks exceeding model limit:", len(over_limit))

print("\nCONTENT TYPE DISTRIBUTION")
print("=" * 60)

print(embedding_chunks_df["content_type"].value_counts())

print("\nSAMPLE SPLIT CHUNKS")
print("=" * 60)

print(
    embedding_chunks_df[
        ["embedding_chunk_id", "original_chunk_id",
         "page_number", "part_number", "token_count"]
    ].head(15).to_string(index=False)
)

assert len(over_limit) == 0, "Some chunks still exceed the model limit!"

print("\nSUCCESS: All embedding chunks fit within the model limit.")
```

```python

# ============================================================
# BUILD HR POLICY RETRIEVAL ENGINE
# ============================================================

import numpy as np
import pandas as pd

# Confirm that vectors and metadata match
assert len(embedding_chunks_df) == len(embedding_vectors)

# Search function
def search_hr_policy(question, top_k=5):

    # Convert employee question into an embedding
    question_vector = embedding_model.encode(
        [question],
        convert_to_numpy=True,
        normalize_embeddings=True
    )

    # Calculate similarity against all policy chunks
    similarity_scores = np.dot(
        embedding_vectors,
        question_vector[0]
    )

    # Get the highest scoring results
    top_indices = np.argsort(similarity_scores)[::-1][:top_k]

    results = []

    for rank, index in enumerate(top_indices, start=1):

        row = embedding_chunks_df.iloc[index]

        results.append({
            "rank": rank,
            "similarity_score": float(similarity_scores[index]),
            "page_number": int(row["page_number"]),
            "chunk_id": row["embedding_chunk_id"],
            "original_chunk_id": row["original_chunk_id"],
            "content_type": row["content_type"],
            "extraction_status": row["extraction_status"],
            "text": row["chunk_text"]
        })

    return pd.DataFrame(results)


print("HR POLICY RETRIEVAL ENGINE CREATED")
print("=" * 60)
print("Available policy embeddings:", len(embedding_vectors))
print("Embedding dimensions:", embedding_vectors.shape[1])
print("Search function: search_hr_policy()")
```

```python
import torch
import gc

from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig
)

# Confirm GPU
assert torch.cuda.is_available(), "GPU is not available!"

# Model name
qwen_model_name = "Qwen/Qwen2.5-3B-Instruct"

# Configure 4-bit quantization
quantization_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True
)

print("Loading Qwen tokenizer...")

qwen_tokenizer = AutoTokenizer.from_pretrained(
    qwen_model_name
)

print("Loading Qwen model...")

qwen_model = AutoModelForCausalLM.from_pretrained(
    qwen_model_name,
    quantization_config=quantization_config,
    device_map="auto"
)

qwen_model.eval()

print("\nMODEL LOADED SUCCESSFULLY")
print("=" * 60)
print("Model:", qwen_model_name)
print("Tokenizer:", qwen_tokenizer.__class__.__name__)
print("GPU:", torch.cuda.get_device_name(0))
print(
    "GPU memory allocated (GB):",
    round(torch.cuda.memory_allocated(0) / (1024**3), 2)
)
```

```python
messages = [
    {
        "role": "system",
        "content": "You are a helpful AI assistant."
    },
    {
        "role": "user",
        "content": "Explain what an HR policy is in two sentences."
    }
]

prompt = qwen_tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)

inputs = qwen_tokenizer(
    prompt,
    return_tensors="pt"
).to(qwen_model.device)

with torch.no_grad():
    outputs = qwen_model.generate(
        **inputs,
        max_new_tokens=120,
        do_sample=False
    )

response = qwen_tokenizer.decode(
    outputs[0][inputs["input_ids"].shape[-1]:],
    skip_special_tokens=True
)

print("QWEN RESPONSE")
print("=" * 60)
print(response)
```

```python
import os
import numpy as np
import pandas as pd
from sentence_transformers import SentenceTransformer

# 1. Define checkpoint locations
checkpoint_dir = "/content/hr_policy_ai_checkpoint"
embeddings_dir = os.path.join(checkpoint_dir, "embeddings")
embedding_model_dir = os.path.join(checkpoint_dir, "embedding_model")

# 2. Load the saved embedding model
print("Loading embedding model...")

embedding_model = SentenceTransformer(
    embedding_model_dir,
    device="cuda" if __import__("torch").cuda.is_available() else "cpu"
)

# 3. Load embedding metadata
print("Loading embedding metadata...")

embedding_chunks_df = pd.read_csv(
    os.path.join(embeddings_dir, "embedding_metadata.csv")
)

# 4. Load precomputed embedding vectors
print("Loading embedding vectors...")

embedding_vectors = np.load(
    os.path.join(embeddings_dir, "hr_policy_embeddings.npy")
)

# 5. Validate restored data
print("\nRESTORATION VALIDATION")
print("=" * 60)

print("Embedding model:", type(embedding_model).__name__)
print("Embedding dimensions:", embedding_model.get_embedding_dimension())
print("Embedding chunks:", len(embedding_chunks_df))
print("Embedding vectors shape:", embedding_vectors.shape)

print("\nChecking alignment...")

assert len(embedding_chunks_df) == len(embedding_vectors), \
    "Mismatch between metadata and embedding vectors!"

assert embedding_vectors.shape[1] == embedding_model.get_embedding_dimension(), \
    "Embedding dimensions do not match!"

assert embedding_chunks_df["chunk_text"].notna().all(), \
    "Some embedding chunks have missing text!"

print("\nSUCCESS: HR retrieval data restored and validated.")
```

```python
def search_hr_policy(question, top_k=5):

    # Convert employee question into an embedding vector
    question_vector = embedding_model.encode(
        [question],
        convert_to_numpy=True,
        normalize_embeddings=True
    )

    # Calculate similarity against stored policy vectors
    similarity_scores = np.dot(
        embedding_vectors,
        question_vector[0]
    )

    # Find the highest-scoring chunks
    top_indices = np.argsort(similarity_scores)[::-1][:top_k]

    results = []

    for rank, index in enumerate(top_indices, start=1):

        row = embedding_chunks_df.iloc[index]

        results.append({
            "rank": rank,
            "similarity_score": float(similarity_scores[index]),
            "page_number": int(row["page_number"]),
            "chunk_id": row["embedding_chunk_id"],
            "original_chunk_id": row["original_chunk_id"],
            "content_type": row["content_type"],
            "text": row["chunk_text"]
        })

    return pd.DataFrame(results)


print("HR POLICY SEARCH FUNCTION RESTORED SUCCESSFULLY.")
```

```python
import torch

def answer_hr_policy(question, top_k=5):

    # STEP 1: Retrieve relevant HR policy passages
    retrieved_results = search_hr_policy(
        question,
        top_k=top_k
    )

    # STEP 2: Prepare policy evidence
    policy_context = ""

    for _, row in retrieved_results.iterrows():

        policy_context += f"""
SOURCE:
Page: {row['page_number']}
Chunk ID: {row['chunk_id']}

POLICY CONTENT:
{row['text']}

----------------------------------------
"""

    # STEP 3: Strict instructions
    system_prompt = """
You are the Internal HR Policy Assistant for I&M Bank Tanzania.

Answer employee questions using ONLY the supplied HR policy evidence.

RULES:
1. Never use general knowledge to answer policy questions.
2. Never invent procedures, rules, deadlines or responsibilities.
3. If the evidence is insufficient, say:

"I could not find sufficient information in the available HR policy
extracts to answer this question reliably. Please contact the
Human Resources Department for clarification."

4. If a question is unrelated to HR policy, explain that you only
assist with questions covered by the HR policy.
5. Cite page numbers for important policy statements.
6. Never invent page numbers.
7. Distinguish between different policy procedures.
8. Give clear, simple answers.

Accuracy is more important than answering every question.
"""

    user_prompt = f"""
EMPLOYEE QUESTION:
{question}

APPROVED HR POLICY EVIDENCE:
{policy_context}

Answer using only the supplied evidence. Include page citations.
"""

    # STEP 4: Create the conversation
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt}
    ]

    # STEP 5: Convert messages into PyTorch input tensors
    inputs = qwen_tokenizer.apply_chat_template(
        messages,
        tokenize=True,
        add_generation_prompt=True,
        return_tensors="pt",
        return_dict=True
    )

    # Move input tensors to the Qwen GPU
    inputs = {
        key: value.to(qwen_model.device)
        for key, value in inputs.items()
    }

    # STEP 6: Generate response
    qwen_model.eval()

    with torch.no_grad():

        output = qwen_model.generate(
            **inputs,
            max_new_tokens=450,
            do_sample=False,
            repetition_penalty=1.1,
            pad_token_id=qwen_tokenizer.eos_token_id
        )

    # STEP 7: Extract only newly generated tokens
    input_length = inputs["input_ids"].shape[1]

    generated_tokens = output[0][input_length:]

    answer = qwen_tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    ).strip()

    return answer


print("HR POLICY ANSWER FUNCTION UPDATED SUCCESSFULLY.")
```
